Top skills for a Data Analyst

### Avg Salary per skill





```python
%%sql

SELECT 
    sd.skills,
    ROUND(AVG(jpf.salary_year_avg::numeric), 2) as avg_salary,
    COUNT(*) as num_jobs
FROM job_postings_fact jpf
JOIN skills_job_dim sjd ON jpf.job_id = sjd.job_id
JOIN skills_dim sd ON sjd.skill_id = sd.skill_id
WHERE jpf.salary_year_avg::numeric IS NOT NULL
GROUP BY sd.skills,jpf.salary_year_avg
HAVING COUNT(*) > 10
ORDER BY jpf.salary_year_avg DESC
LIMIT 10;
```

#### skills that appear often ... sql, python

```python
%%sql

SELECT sd1.skills as skill1, 
    sd2.skills as skill2,
    COUNT(*) as co_occurrence_count
FROM skills_job_dim sjd1
JOIN skills_job_dim sjd2 
    ON sjd1.job_id = sjd2.job_id 
    AND sjd1.skill_id < sjd2.skill_id
JOIN skills_dim sd1 ON sjd1.skill_id = sd1.skill_id
JOIN skills_dim sd2 ON sjd2.skill_id = sd2.skill_id
GROUP BY sd1.skills, sd2.skills
ORDER BY co_occurrence_count DESC
LIMIT 20;

```

%% skills trending remote work vs on site

```python
%%sql


SELECT sd.skills,
    jpf.job_work_from_home,
    COUNT(*) as demand_count
FROM job_postings_fact jpf
JOIN skills_job_dim sjd ON jpf.job_id = sjd.job_id
JOIN skills_dim sd ON sjd.skill_id = sd.skill_id
GROUP BY sd.skills, jpf.job_work_from_home
ORDER BY sd.skills, jpf.job_work_from_home DESC
LIMIT 20;
```
